```python
# Photon rate at a known magnitude.
rate = 3.0E+04    # photons per second
mag0 = 13.0

# Rate for another magnitude.
other_mag = 15.5
other_rate = 10.0**(0.4 * (other_mag - mag0)) * rate
print('A star of magnitude %5.2f will deliver %6.1e photons/second.'
      % (other_mag, other_rate))
```
